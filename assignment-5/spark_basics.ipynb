```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("SparkAssignment") \
    .getOrCreate()

print("Spark Session Created")
```

```python
df = spark.read.csv(
    "sparkdataset.csv",
    header=True,
    inferSchema=True
)

df.show()
```

```python
print(df.columns)

df.printSchema()
```

```python
df = df.dropDuplicates()

print("Rows after removing duplicates:")
print(df.count())
```

```python
df = df.fillna({
    "age": 0,
    "category": "Unknown",
    "salary": 0
})

df.show()
```

```python
df.filter(df.age > 25).show()
```

```python
df.filter(df.category == "IT").show()
```

```python
df.filter(df.region == "North").show()
```

```python
df = df.withColumnRenamed(
    "category",
    "department"
)

df.show()
```

```python
from pyspark.sql.functions import col

df = df.withColumn(
    "salary",
    col("salary").cast("int")
)

df.printSchema()
```

```python
from pyspark.sql.functions import *
```

```python
df.select(count("*")).show()
```

```python
df.select(avg("salary")).show()
```

```python
df.select(min("salary")).show()
```

```python
df.select(max("salary")).show()
```

```python
df.groupBy("department") \
  .agg(
      count("*").alias("employee_count"),
      avg("salary").alias("avg_salary"),
      sum("salary").alias("total_salary")
  ) \
  .show()
```

```python
df.groupBy("department").count().show()
```

```python
pipeline_df = (
    df.dropDuplicates()
      .fillna({
          "age": 0,
          "salary": 0,
          "department": "Unknown"
      })
)

pipeline_df.groupBy("department") \
           .agg(
               count("*").alias("employees"),
               avg("salary").alias("avg_salary")
           ) \
           .show()
```

```python
result = pipeline_df.groupBy("department") \
                    .agg(
                        count("*").alias("employees"),
                        avg("salary").alias("avg_salary")
                    )

result.coalesce(1).write.mode("overwrite").csv("results")
```
